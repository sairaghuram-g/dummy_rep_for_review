```python
import numpy as np
```

```python
x = [.25, .75]
H=0
for p in x:
    H -= p*np.log(p)

print(H)
```

##Cross Entropy

```python
p = [1.0, 0.0]
q = [.25,.75]

H=0
for i in range(len(p)):
    H -= p[i]*np.log(q[i])

print(H)
```

##Pytorch

```python
import torch
import torch.nn.functional as F

p_t = torch.tensor(p)
q_t = torch.tensor(q)

F.binary_cross_entropy(q_t,p_t)
```
