```python
import numpy as np
import torch
import torch.nn as nn
```

```python
widenet = nn.Sequential(
    nn.Linear(2,4),
    nn.Linear(4,3),
)

deepnet = nn.Sequential(
    nn.Linear(2,2),
    nn.Linear(2,2),
    nn.Linear(2,3),
)

print(widenet)
print(deepnet)
```

```python
#checkout parameters
for p in deepnet.named_parameters():
    print(p)
```

```python
#simple way to see model summary
from torchsummary import summary
widenet = widenet.to('cuda')
summary(widenet, (1, 2), device='cuda')
```
